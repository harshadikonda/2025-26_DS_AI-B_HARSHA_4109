```python
#bubble sort
n=int(input("enter the number of elements"))
arr=[]
for i in range(n):
  arr.append(int(input(f"enter element{i+1}:")))
print("\n original array:",arr)
for i in range(n-1):
  for j in range(n-1-i):
    if arr[j]>arr[j+1]:
      arr[j],arr[j+1]=arr[j+1],arr[j]
print("sorted array:",arr)
```
